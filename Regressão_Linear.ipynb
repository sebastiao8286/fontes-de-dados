```python
import pandas as pd
url = "https://raw.githubusercontent.com/ybifoundation/Dataset/main/Salary%20Data.csv"
df = pd.read_csv(url)
display(df.head())
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Calculate the coefficients for the linear trend line (y = mx + c)
m, c = np.polyfit(df['Experience Years'], df['Salary'], 1)

# Generate y values for the trend line based on the calculated coefficients
trend_line = m * df['Experience Years'] + c

# Create the scatter plot
plt.scatter(df['Experience Years'], df['Salary'], label='Data Points')

# Plot the trend line
plt.plot(df['Experience Years'], trend_line, color='red', label=f'Trend Line (y = {m:.2f}x + {c:.2f})')

plt.xlabel('Experience Years')
plt.ylabel('Salary')
plt.title('Salary vs Experience Years with Trend Line')
plt.legend()
plt.show()
```

```python
print(f"Interceptor (c): {c:.2f}")
print(f"Coef (m): {m:.2f}")
```
